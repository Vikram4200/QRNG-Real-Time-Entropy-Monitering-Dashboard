```python
# Cell 4: QRNG Workflow Diagram using Matplotlib

import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

def draw_box(ax, x, y, text, color):
    box = FancyBboxPatch(
        (x, y), 2.6, 0.8,
        boxstyle="round,pad=0.08",
        linewidth=1.5,
        edgecolor="black",
        facecolor=color
    )
    ax.add_patch(box)
    ax.text(x + 1.3, y + 0.4, text, ha="center", va="center", fontsize=10)

def draw_arrow(ax, x1, y1, x2, y2):
    arrow = FancyArrowPatch(
        (x1, y1), (x2, y2),
        arrowstyle="->",
        mutation_scale=15,
        linewidth=1.5,
        color="black"
    )
    ax.add_patch(arrow)

fig, ax = plt.subplots(figsize=(14, 3))

steps = [
    ("2-Qubit\nCircuit", "#dbeafe"),
    ("Quantum\nSuperposition", "#dcfce7"),
    ("Measurement", "#fef3c7"),
    ("Raw\nBitstream", "#fde2e2"),
    ("Correction\nMethods", "#e9d5ff"),
    ("Randomness\nTests", "#cffafe"),
    ("EHI\nDashboard", "#fbcfe8")
]

x_positions = [0, 3, 6, 9, 12, 15, 18]

for i, (text, color) in enumerate(steps):
    draw_box(ax, x_positions[i], 1, text, color)
    if i < len(steps) - 1:
        draw_arrow(ax, x_positions[i] + 2.6, 1.4, x_positions[i + 1], 1.4)

ax.set_xlim(-0.5, 21)
ax.set_ylim(0, 3)
ax.axis("off")
plt.title("QRNG Workflow", fontsize=14, fontweight="bold")
plt.show()
```
