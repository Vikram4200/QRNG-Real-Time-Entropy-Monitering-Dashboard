```python
%matplotlib inline
from qiskit import QuantumCircuit, transpile # 
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
# circuit program start from here
qc = QuantumCircuit(1)
qc.h(0)
# qc.cx(1,0)
qc.measure_all()
# simulation cmd
sim = AerSimulator()

compiled = transpile(qc,sim)
result = sim.run(compiled, shots=1024).result()
counts = result.get_counts()
# qc.x(2)
# qc.y(1)
# qc.z(0)
# qc.cx(1,0)
# qc.p(3.14,2)
# qc.swap(0,1)
# qc.ccx(0,1,3) # control qubits

#output
qc.draw(output='mpl')
print("Measurement counts : ",counts)
plot_histogram(counts)
plt.show() 
```

```python
# ============================================
# EXPERIMENT 1 : X-GATE ON SINGLE QUBIT
# ============================================

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1,1)

# Apply X Gate
qc.x(0)

# Measurement
qc.measure(0,0)

# Draw circuit
print(qc.draw())

# Simulation
simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()

counts = result.get_counts()

print("Output Counts:", counts)

# Histogram
plot_histogram(counts)
plt.show()
```

```python
%matplotlib inline
from qiskit import QuantumCircuit, transpile # 
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
from qiskit.quantum_info import Statevector  # state 
from qiskit.visualization import plot_bloch_multivector
# circuit program start from here
qc = QuantumCircuit(3) # 3qubit
qc.h(0)
qc.h(1)
qc.h(2)

# qc.ccx(0,1,2)
# state = Statevector.from_instruction(qc)
# plot_bloch_multivector(state)
qc.measure_all()
# simulation cmd
sim = AerSimulator()

compiled = transpile(qc,sim)
result = sim.run(compiled, shots=1).result()
counts = result.get_counts()

binary_number = list(counts.keys())[0]   #extract binary output
decimal_number = int(binary_number, 2)  # convert to binary to decimal 

#output
qc.draw(output='mpl')
print("Random Binary Number:", binary_number)
print("random Decimal Number :", decimal_number)
# print("Measurement counts : ",counts)
plot_histogram(counts)
plt.show() 
```
